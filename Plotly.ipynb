```python
import plotly.graph_objects as go
import plotly.express as px
#using graph object

fig= go.Figure()
#scatter_plot
# add_trace >> adding data
fig.add_trace(go.Scatter(x=[1,2,3,4,5], y=[1,7,8,2,6], mode= 'markers'))
```

```python
#Line Chart
fig= go.Figure()
fig.add_trace(go.Scatter(x=[1,2,3,4,5], y=[1,7,8,2,6], mode= 'lines'))
```

```python
#Scatter Plot
fig= go.Figure()
fig.add_trace(go.Scatter(x=tips.total_bill, y=tips.tip, mode= 'markers'))
```

```python
#Scatter Plot
fig= go.Figure()
fig.add_trace(go.Scatter(x=tips.total_bill, y=tips.tip, mode= 'lines'))
```

```python
#Bar Chart
fig= go.Figure()
fig.add_trace(go.Bar(x=[1,2,3,4,5], y=[3,8,5,2,6]))
```

```python
#Histogram
fig=go.Figure()
fig.add_trace(go.Histogram(x=tips['total_bill']))
```

```python
fig= go.Figure()
fig.add_trace(go.Scatter(x=tips.total_bill, y=tips.tip, mode= 'markers', marker_size=tips['size']))
```

```python
fig= go.Figure()
fig.add_trace(go.Scatter(x=tips.total_bill, y=tips.tip, mode= 'markers', marker_size=12*tips['size']))
```

```python
#using plotly express
#Scatter Plot
x=[1,2,3,4]
y=[10,13,15,18]
fig= px.scatter(x,y)
fig.show()
```

```python
#Line Plot

fig=px.line(x,y)
fig.show()
```

```python
x=['A','B','C']
y=[10,5,6]
fig = px.bar(x=x,y=y)
fig.show()
```

```python
#Histogram
x=[5,10,3,4,5,5,6,6,10,10,3,11,11,12,3,4]
fig=px.histogram(x)
fig.show()
```

```python
#Box Plot
x=['A','B','C']
Y=[5,10,3,25,12,33,100]
fig=px.box(x,y)
fig.show()
```

```python
#Plottting Heatmap
data=[[1,2,3],[4,5,6],[7,8,9]]
fig= px.imshow(data)
fig.show()
```

```python
#Bubble Chart
x=[1,2,3,4]
y=[5,8,12,3]
fig=px.scatter(x,y,size=[30,10,50,100])
fig.show()
```

```python
#3D Chart
x=[1,2,3,4]
y=[5,8,12,3]
z=[3,4,5,6]
fig= px.scatter_3d(x=x,y=y,z=z)
fig.show()
```

```python
#Contour Plot
x=[1,2,3,4,3,23,80]
y=[5,8,12,3,12,35,60]
fig=px.density_contour(x,y)
fig.show()
```
