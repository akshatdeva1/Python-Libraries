```python
x=[1,2,3,4,5]
y1=[5,1,7,8,2]
y2=[7,8,3,5,1]

plt.plot(x,y1, color="red", marker="o", linestyle = "--", linewidth=3, markersize=6, label="line y1 as tata motors")
plt.plot(x,y2, color="blue", marker="o", linestyle = "--", linewidth=3, markersize=6, label="line y2 as tata power")

plt.title("Multiple lines in one plot")

plt.xlabel("Days")
plt.ylabel("Stock Price")
plt.legend()
plt.show()
```

```python
#Histogram >> to understand continuous numerical data

data=np.random.randn(1000)   #by default 10 bins are created
plt.hist(data)
plt.show()
```

```python
#Stacked Histogram >> two histogrsm together

data1=np.random.normal(0,1,5000)
data2=np.random.normal(3,2,5000)
print(data1)
print(data2)

plt.hist([data1,data2],bins=20,color=['r','b'], label=['dist1','dist2'])
plt.xlabel("Values")
plt.ylabel("frequency")
plt.title("Stacked Historgram")
plt.show()
```

```python
#Pie Chart >> is usee to undestand the percentage distribution of data

langs=['C', 'Python','Java']
students=[20,100,40]

plt.pie(students,labels=langs, autopct= "%1.1f%%")
plt.show()
```

```python
#focus on one segment
explode=(0.0,0.1,0.1)
langs=['C', 'Python','Java']
students=[20,100,40]

plt.pie(students,labels=langs, autopct= "%1.1f%%",explode=explode)
plt.show()
```

```python
explode=(0.0,0.1,0.1)
langs=['C', 'Python','Java']
students=[20,100,40]

plt.pie(students,labels=langs, autopct= "%1.1f%%",explode=explode, shadow=True)
plt.show()
```

```python
#Area Plot/Stacked Area Plot >> Very Important >> Variable changed overtime

years=[2021,2018,2022,2023,2017]
category1=[20,25,30,28,35]
category2=[10,15,20,18,25]
category3=[5,8,10,9,15]

plt.stackplot(years,category1,category2,category3, labels=["category1","category2","category3"], colors=["skyblue","lightgreen","orange"])
plt.xlabel("years")
plt.ylabel("Values")
plt.title("Area Plot")
plt.legend()
plt.show()
```
