```python
#numerical data plotting using seaborn
#relplot has options of plotting both categorical and numerical data
#relplot>> relstionship between two variables
#relplot>> relationship plot

#scatterplot, Lineplot
```

```python
#how tip is varying wrt total bill >> scatterplot/by default relplot make scatter plot

sns.relplot(x='total_bill', y='tip', data=df) #relationship b/w two variables, by default it gives scatter plot

#kind='line' >> it will change the plot
#insight >> as total billl increases, tip increases
#customer who smoked gave theh highest tip
```

```python
sns.relplot(x='total_bill', y='tip', data=df, hue='smoker')
```

```python
sns.relplot(x='total_bill', y='tip', data=df, hue='size') 

#hue is giving the different color coding for the column value that you passed
#hue gives the different color shades
```

```python
sns.relplot(x='total_bill', y='tip', data=df, style='size') 
#style give different representation of column value
```

```python
sns.relplot(x='total_bill', y='tip', data=df, style='smoker', hue='time') 
```

```python
#plot between data and total_bill, day is a categorical column
#for categorical column >> catplot
sns.catplot(x='day', y='total_bill', data=tips)
```

```python
#insights>> many people arrives on weekend as compare to weekdays
#in cat plot, few datapoints are overlapping>> to see it separately use swarm plot
```

```python
#Swarm Plot
sns.catplot(x='day', y='total_bill', data=tips,kind='swarm')
```

```python
#BoxPlot >> to see outliers
sns.catplot(x='day', y='total_bill', data=tips, kind='box')
```

```python
#Voilin Plot
sns.catplot(x='day', y='total_bill', data=tips, kind='violin')
```

```python
#Boxen Plot >> made from box plot itself

sns.catplot(x='day', y='total_bill', data=tips, kind='boxen')
```

```python
sns.barplot(x='day', y='total_bill',data= tips)
```

```python
#Joint plot/Bivariate plot(means two plots)
sns.jointplot(x=tips.total_bill, y=tips.tip)    #draw a plot of two variables with distribution
```

```python
#insights >> understand relationship and also distribution of two variables
```

```python
sns.jointplot(x=tips.total_bill, y=tips.tip, kind='reg')   #reg >> regression >> to represent relationship
```

```python
#kernel density estimator(KDE)>>density of data in particular interval
```

```python
sns.jointplot(x=tips.total_bill, y=tips.tip, kind='hex') 
```

```python
sns.jointplot(x=tips.total_bill, y=tips.tip, kind='kde')    #Kernel density estimator >> smoothens the histogram to understand the density
```
