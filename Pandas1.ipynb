```python
#to check the execution/ run time of a code

start= pd.Timestamp.now()
for i in range(1000):
    pass
end= pd.Timestamp.now()
print("Time Taken for the execution of code", end-start)
```

```python
#another way

import time
start= time. time()
start
for i in range(1000):
    pass
end= time.time()
print("Time taken for the execution of the code in seconds", end-start)
```

```python
#One more way

import time
start= time.perf_counter()
for i in range(1000):
    pass
end= time.perf_counter()
print("Time taken for the execution of the code in seconds", end-start)
```
